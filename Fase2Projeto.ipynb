```python
lista_modulos = [ #Lista dos modulos para pouso, com id, tipo, prioridade, combustivel, massa (peso), criticidade da carga e horario para pouso
{
        "id_modulo" :"MOD1",
        "tipo":"habitação",
        "prioridade": 1,
        "combustivel": 18.5,
        "massa_ton": 18.5,
        "criticidade_carga": "Alta",
        "horario_orbita": "14:30"
    },
    {
        "id_modulo" :"MOD2",
        "tipo":"energia",
        "prioridade": 1,
        "combustivel": 14.2,
        "massa_ton": 14.2,
        "criticidade_carga": "Alta",
        "horario_orbita": "14:35"

    },
    {
        "id_modulo" :"MOD3",
        "tipo":"suporte_medico",
        "prioridade": 2,
        "combustivel": 15.0,
        "massa_ton": 9.8,
        "criticidade_carga": "Alta",
        "horario_orbita": "14:30"
    },
    {
        "id_modulo" :"MOD4",
        "tipo":"laboratorio",
        "prioridade": 4,
        "combustivel": 12.0,
        "massa_ton": 12.0,
        "criticidade_carga": "Média",
        "horario_orbita": "14:30"
    },
    {
        "id_modulo" :"MOD5",
        "tipo":"logistica",
        "prioridade": 3,
        "combustivel": 22.0,
        "massa_ton": 22.0,
        "criticidade_carga": "Média",
        "horario_orbita": "14:30"
    },
    {
        "id_modulo" :"MOD6",
        "tipo":"mineração",
        "prioridade": 5,
        "combustivel": 7.1,
        "massa_ton": 25.1,
        "criticidade_carga": "Baixa",
        "horario_orbita": "14:30"
    }
]

fila_pouso = [] #Fila para pouso onde os modulos serão adicionados
for modulo in lista_modulos:
    fila_pouso.append(modulo) #Inserir cada modulo na fila

pilha_emergencia = [] #Pilha para adicionar os modulos que estão com pouco combustivel

def buscar_menor_combustivel(lista): #Localiza o modulo com menor combustivel
    if not lista:
        return None
    menor = lista[0]
    for modulo in lista:
        if modulo["combustivel"] < menor["combustivel"]:
            menor = modulo
    return menor


def ordenar_fila_pouso(fila): #Reorganiza a fila em ordem de pouso

    n = len(fila)
    for i in range(n):
        for j in range(0, n - i - 1):
            trocar = False

            # Critério 1: Menor número de prioridade vem antes
            if fila[j]["prioridade"] > fila[j + 1]["prioridade"]:
                trocar = True
            # Critério 2: Desempate por menor combustível
            elif fila[j]["prioridade"] == fila[j + 1]["prioridade"]:
                if fila[j]["combustivel"] > fila[j + 1]["combustivel"]:
                    trocar = True

            if trocar:
                fila[j], fila[j + 1] = fila[j + 1], fila[j]


def avaliar_autorizacao_pouso(modulo, clima_favoravel, pista_livre):
    """
    Reflete as condições lógicas booleanas para autorização do pouso.
    Regra: (Pista Livre AND Clima OK) OR (Combustível Crítico AND Pista Livre)
    """
    combustivel_critico = modulo["combustivel"] < 10.0

    prioridade_maxima = modulo["prioridade"] == 1

    if not pista_livre:
        return False, "Pouso Negado: Pista de pouso ocupada."
    elif not clima_favoravel:
        if combustivel_critico:
            return True, "ALERTA: Pouso de emergência autorizado devido a combustível crítico!"
        else:
            return False, "Pouso Negado: Condições atmosféricas desfavoráveis."
    else:
        if combustivel_critico or prioridade_maxima:
            return True, "Pouso Prioritário Autorizado."
        else:
            return True, "Pouso Regular Autorizado."


print("FILA DE POUSO INICIAL \n")
for m in fila_pouso:
    print(f"ID: {m['id_modulo']} | Tipo: {m['tipo']} | Prioridade: {m['prioridade']} | Combustível: {m['combustivel']}L")

# Reorganizando a fila de pouso
ordenar_fila_pouso(fila_pouso)

print("FILA DE POUSO REORGANIZADA (ORDEM DE PRIORIDADE) \n")
for m in fila_pouso:
    print(f"ID: {m['id_modulo']} | Tipo: {m['tipo']} | Prioridade: {m['prioridade']} | Combustível: {m['combustivel']}L")

# Simulação de processamento da fila
clima_ok = True
pista_livre = True

print("PROCESSAMENTO DE AUTORIZAÇÃO DE POUSO")
while len(fila_pouso) > 0:
    # Remoção do primeiro elemento da fila (FIFO)
    modulo_atual = fila_pouso.pop(0)

    autorizado, motivo = avaliar_autorizacao_pouso(modulo_atual, clima_ok, pista_livre)
    print(f"Processando '{modulo_atual['tipo']}': {motivo}")

    if not autorizado:
        # Caso o pouso seja negado, o módulo é enviado para a pilha de contingência
        pilha_emergencia.append(modulo_atual)
```
